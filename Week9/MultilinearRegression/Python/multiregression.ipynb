## Import libraries and load the dataset

```python
import pandas as pd
import seaborn as sns
import statsmodels.api as sm

# Load dataset, .dropna() is a shortcut to remove mising rows or columns
mpg = sns.load_dataset("mpg").dropna()

# Display variable names and info
mpg.info()
```

## Define variables that will be used for the equation

```python
# Define predictors
X = mpg[["horsepower", "weight"]]

# Define target
y = mpg["mpg"]

# Add intercept
X = sm.add_constant(X)
```

## Implement the multi-regression model

```python
# Fit regression model
model = sm.OLS(y, X).fit()

# Print summary
print(model.summary())

# Extract R-squared
r2 = model.rsquared

# Convert to percentage accuracy
accuracy_percent = r2 * 100

print("\nModel Accuracy (R² as %):", round(accuracy_percent,2), "%")
```
