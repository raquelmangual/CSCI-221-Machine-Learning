```python
# Import libraries
import seaborn as sns
import pandas as pd
```

```python
# Load dataset
tips = sns.load_dataset("tips")
```

```python
# Display summary statistics
print("\nSummary Statistics")
print(tips.describe())

# Display column names, data types, and non-null counts
print("\nDataset Details")
tips.info()
```

```python
# Calculate quartiles
Q1 = tips["total_bill"].quantile(0.25)
Q3 = tips["total_bill"].quantile(0.75)
```

```python
# Compute IQR
IQR = Q3 - Q1

# Determine outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# Identify outliers
outliers = tips[(tips["total_bill"] < lower_bound) | (tips["total_bill"] > upper_bound)]

# Count outliers
print("Number of outliers:", len(outliers))

# List outliers
print("Outlier rows:")
print(outliers)
```

```python
# Remove outliers
tips_clean = tips[(tips["total_bill"] >= lower_bound) & (tips["total_bill"] <= upper_bound)]

# Compare dataset sizes
print("Original dataset size:", tips.shape)
print("Dataset size after removing outliers:", tips_clean.shape)

# The dataset sizes can also be compared with other methods, such as .info()
```
