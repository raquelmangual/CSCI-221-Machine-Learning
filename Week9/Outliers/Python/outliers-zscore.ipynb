```python
import seaborn as sns
import numpy as np
from scipy import stats

# Load dataset
tips = sns.load_dataset("tips")
```

```python
# Calculate z-scores
tips["z_score"] = stats.zscore(tips["total_bill"])

# Identify outliers
outliers = tips[abs(tips["z_score"]) > 3]
```

```python
# Count outliers
print("Number of outliers:", len(outliers))

# List outliers
print("Outlier rows:")
print(outliers)
```

```python
# Remove outliers
tips_clean = tips[abs(tips["z_score"]) <= 3]

# Compare dataset sizes
print("Original dataset size:", tips.shape)
print("Dataset size after removing outliers:", tips_clean.shape)
```
